```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
data = {
    'Employee_ID': range(1, 21),
    'Salary': [25000, 28000, 30000, 32000, 35000, 37000, 40000, 42000, 45000, 47000, 48000, 50000, 52000, 55000, 58000, 60000, 62000, 65000, 180000, 220000]
}
df = pd.DataFrame(data)
print("EMPLOYEE SALARY DATASET")
display(df)
Q1 = df['Salary'].quantile(0.25)
Q3 = df['Salary'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = df[(df['Salary'] < lower_bound) | (df['Salary'] > upper_bound)]
print("\nDESCRIPTIVE STATISTICS")
display(df['Salary'].describe())
print("\nOUTLIER DETECTION RESULTS")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Total Records:", len(df))
print("Number of Potential Outliers:", len(outliers))
print("\nIdentified Potential Outliers:")
display(outliers)
df['Status'] = np.where(df['Salary'].between(lower_bound, upper_bound), 'Normal', 'Potential Outlier')
plt.figure(figsize=(10, 5))
plt.boxplot(df['Salary'], patch_artist=True, showmeans=True)
plt.title('Box Plot for Employee Salary Outlier Detection')
plt.ylabel('Salary (₹)')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()
plt.figure(figsize=(10, 5))
normal = df[df['Status'] == 'Normal']
outlier = df[df['Status'] == 'Potential Outlier']
plt.scatter(normal['Employee_ID'], normal['Salary'], label='Normal Values', marker='o')
plt.scatter(outlier['Employee_ID'], outlier['Salary'], label='Potential Outliers', marker='x', s=120)
plt.title('Employee Salary: Normal Values vs Potential Outliers')
plt.xlabel('Employee ID')
plt.ylabel('Salary (₹)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
print("\nFINAL SUMMARY")
print("Normal Records:", len(normal))
print("Potential Outliers:", len(outlier))
print("Outlier Salaries:", outlier['Salary'].tolist())
print("\nCONCLUSION")
print("The box plot and IQR method identified unusually high salary values.")
print("Potential outliers should be investigated before deciding whether to remove them.")
display(df)
```
